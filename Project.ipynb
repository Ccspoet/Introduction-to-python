### Project 1. Student Registration Number 
AIMS-Senegal would like to automate the registration of its students by assigning each of them
a unique registration number. The mechanism below will be used to generate this number.
1. Primary Fields: Each student’s registration number is composed of three fields:
• Field 1: Extract the first three consonants of the last name (in uppercase).
• Field 2: Extract the first three letters of the first name (in uppercase).
• Field 3: A three-digit sequential counter representing the order in which students
with the same first two fields register (starting at 001).
If the last name does not contain three consonants, pad the missing ones with “X”. If the
first name has fewer than three letters, also pad with “X”. The sequential counter must
always be exactly three digits long (e.g., 001, 002, . . . , 999).
2. Uniqueness Requirement: Two students with the same first three consonants of last
name and first name initials must still get unique numbers. For instance:
• Gnansounou Marcel ⇒ GNNMAR001
• Gnonas Marcellin (registered later) ⇒ GNNMAR002
3. Additional Requirements (Complexity):
• Case-insensitivity: The program must work regardless of whether the input is
lowercase, uppercase, or mixed case.
• Accents & special characters: Normalize accented characters (é → E, ç → C, ü
→ U, etc.).
• Hyphenated and compound names: Ignore spaces and hyphens when extracting
consonants or letters (e.g., “Jean-Pierre” or “Van der Merwe”).
• Persistent Tracking: Simulate that your program remembers how many students
have already been registered with the same prefix so that subsequent calls generate
the next available number (using an in-memory dictionary).
• Validation: Ensure that names contain only alphabetic characters, spaces, or hy-
phens; reject invalid inputs (numbers, symbols) with an error message.
Your Task: Write a pure Python program (no external libraries) that:
1. Accepts as input: last name and first name of a student.

2. Generates the student’s registration number according to the rules above.
3. Keeps track of already assigned numbers during execution so that new students receive
the correct sequence number.
4. Prints the final registration number.
You might want to run a test using the AIMS-SN 2025 class list.

```python
# Cette fonction nous permet de savoir si le mot de passe contient au moins une lettre majuscule...
# Ellle retourne 1 si le mot de passe contient au moins une lettre majuscule et 0 sinon... 
# Elle prend en entré le mot de passe...

def uppercase_test(pass_word):
    a = 0
    for i in pass_word:
        if i.isupper():
           a += 1 
    if a != 0:
        return 1
    else:
        return 0

####### Example test #######

pass_word = "sfj%sxn"
uppercase_test(pass_word)
```

```python
# Cette fonction nous permet de tester la taille d'un mot de passe...
# Elle retourne 1 si la taille du mot de passe est compris entre 8 inclus et 64 inclus et 0 sinon...
# Elle prend en entré le mot de passe...

def test_password_length(pass_word): 
    if 8 <= len(pass_word) <= 64:
        return 1
    else:
        return 0

####### Example test #######

pass_word = "sfjrfsedfdxfdsres%sxn"
print(f"Nombre de caractère du pass_word: {len(pass_word)}")
test_password_length(pass_word)
```

```python
# Cette fonction nous permet de savoir si le mot de passe contient au moins une lettre minuscule...
# Ellle retourne 1 si le mot de passe contient au moins une lettre minuscule et 0 sinon... 
# Elle prend en entré le mot de passe...

def lowercase_test(pass_word):
    a = 0
    for i in pass_word:
        if i.islower():
           a += 1 
    if a != 0:
        return 1
    else:
        return 0

####### Example test #######

pass_word = "sfj%sxn"
lowercase_test(pass_word)
```

```python
# Cette fonction nous permet de savoir si le mot de passe contient au moins un chiffre...
# Ellle retourne 1 si le mot de passe contient au moins un chiffre et 0 sinon... 
# Elle prend en entré le mot de passe...

def digit_heads(pass_word):
    a = 0
    for i in pass_word:
        if i.isdigit():
           a += 1 
    if a != 0:
        return 1
    else:
        return 0

####### Example test #######

pass_word = "sfj%s2nsda"
digit_heads(pass_word)
```

```python
# Cette fonction nous permet de savoir si le mot de passe contient un espace...
# Ellle retourne 1 si le mot de passe contient un espace et 0 sinon... 
# Elle prend en entré le mot de passe...

def test_not_espace(pass_word):
    if " " in pass_word:
        return 0
    else:
        return 1

####### Example test #######

pass_word = "sfjs 2@nsda"
test_not_espace(pass_word)
```

```python
# Cette fonction nous permet de savoir si le mot de passe contient au moins un caractère spécial...
# Ellle retourne 1 si le mot de passe contient au moins un caractère spécial et 0 sinon... 
# Elle prend en entré le mot de passe...

def caracteres_speciaux(pass_word):
    ref = "!@#$%ˆ&*()_+=[]{};:'¨,<.>/?|/~"
    a = 0
    for i in pass_word:
        if i in ref:
            a += 1
    if a != 0:
        return 1
    else:
        return 0
        
####### Example test #######

pass_word = "sfjs2@nsda"
caracteres_speciaux(pass_word)
```
